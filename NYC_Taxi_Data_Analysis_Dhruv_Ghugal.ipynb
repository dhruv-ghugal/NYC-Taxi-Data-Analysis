#Phase B — Data Cleaning

```python
clean_df=main.copy()
clean_df
```

```python
#Task 34
print("Missing values in clean_df:",clean_df.isnull().sum())
```

```python
#Task 35
duplicate_count = clean_df.duplicated().sum()

print("Duplicate records:",duplicate_count)

if duplicate_count > 0:
    print('Need to remove Duplicates')
    clean_df=clean_df.drop_duplicates()
    print("Duplicates removed.")
else:
    print("No duplicate found.")
```

```python
#Task 36
#For pickup (dtype('<M8[us]')) into datetime64[us] we have used -->pd.to_datetime
clean_df['tpep_pickup_datetime'] = pd.to_datetime(clean_df['tpep_pickup_datetime'])


#For Dropoff into datetime64[us] we have used -->pd.to_datetime
clean_df['tpep_dropoff_datetime'] = pd.to_datetime(clean_df['tpep_dropoff_datetime'])

print(clean_df[['tpep_pickup_datetime', 'tpep_dropoff_datetime']].dtypes)
```

```python
#Task 37
invalid_passenger = clean_df[clean_df['passenger_count'] <= 0]

print("Invalid passenger-count records:", len(invalid_passenger))
print(invalid_passenger[['passenger_count','trip_distance','tpep_pickup_datetime','tpep_dropoff_datetime']].head(10))

#Here we'll make them all as missing value[NaN]
#using np.nan
clean_df.loc[clean_df['passenger_count'] <= 0,'passenger_count'] = np.nan
```

```python
# Task 38
invalid_distance = clean_df[clean_df['trip_distance'] <= 0]

print("Zero/negative distance records:", len(invalid_distance))
print(invalid_distance[['trip_distance','tpep_pickup_datetime','tpep_dropoff_datetime','passenger_count'] ].head(10))
```

```python
# Task 39
invalid_time = clean_df[clean_df['tpep_dropoff_datetime'] < clean_df['tpep_pickup_datetime']]

print("Drop-off before pickup records:", len(invalid_time))

print(invalid_time[
['tpep_pickup_datetime','tpep_dropoff_datetime','trip_distance']].head(10))


#This data will be removed as that is suspicous
clean_df = clean_df[clean_df['tpep_dropoff_datetime'] >=clean_df['tpep_pickup_datetime']].copy()
```

```python
# Task 40
clean_df['trip_duration_minutes'] = ((clean_df['tpep_dropoff_datetime'] -clean_df['tpep_pickup_datetime']).dt.total_seconds() / 60)

print(clean_df['trip_duration_minutes'].describe())
print('-'*200)

#Showing lonmgest trip durations
print("\nLongest trip durations:")
print(clean_df.nlargest(10,'trip_duration_minutes')[
        ['tpep_pickup_datetime',
         'tpep_dropoff_datetime',
         'trip_duration_minutes',
         'trip_distance']])
```

## Task 41 - Cleaning Decisions

- Created a separate copy of the original dataset as `clean_df`.
- No duplicate records were found.
- Passenger counts less than or equal to zero were identified as suspicious and converted to missing values (`NaN`).
- Zero and negative trip distances were investigated but not automatically removed.
- Records where drop-off occurred before pickup were identified as suspicious and removed.
- Extremely long trip durations were investigated as potential data-quality issues.

#Feature Engineering

```python
# Feature 1 - Trip Duration
clean_df['trip_duration_minutes'] = (
    (clean_df['tpep_dropoff_datetime'] -clean_df['tpep_pickup_datetime'])
    .dt.total_seconds() / 60)

print(clean_df[['tpep_pickup_datetime',
                'tpep_dropoff_datetime',
                'trip_duration_minutes']].head())
```

```python
# Feature 2 - Pickup Date

clean_df['pickup_date'] = clean_df['tpep_pickup_datetime'].dt.date
clean_df[['tpep_pickup_datetime', 'pickup_date']].head()
```

```python
# Feature 3 - Pickup Hour

clean_df['pickup_hour'] = (clean_df['tpep_pickup_datetime'].dt.hour)

print(clean_df[['tpep_pickup_datetime',
                'pickup_hour']].head())
```

```python
# Feature 4 - Day of Week

clean_df['day_of_week'] = (clean_df['tpep_pickup_datetime'].dt.day_name())

print(clean_df[['tpep_pickup_datetime',
                'day_of_week']].head(10))
```

```python
# Feature 5 - Weekend or WeekDay

clean_df['is_weekend'] = (clean_df['tpep_pickup_datetime'].dt.dayofweek >= 5)

print(clean_df[['day_of_week',
                'is_weekend']].head(10))
```

```python
# Feature 6 - Average Speed
#Average Speed = Trip Distance / Trip Duration in hours
clean_df['average_speed_mph'] = np.where(clean_df['trip_duration_minutes'] > 0,
    clean_df['trip_distance'] /(clean_df['trip_duration_minutes'] / 60),np.nan)

print(clean_df[['trip_distance',
                'trip_duration_minutes',
                'average_speed_mph']].head())
```

```python
# Feature 7 - Distance Category

clean_df['distance_category'] = pd.cut(
    clean_df['trip_distance'],
    bins=[-np.inf, 3, 10, np.inf],
    labels=['Short', 'Medium', 'Long']
)

print(clean_df[['trip_distance',
                'distance_category']].head())

print('This gives distance groups for comparison and visualization')
```

```python
# Feature 8 - Duration Category
# Very Short: <= 5 minutes
# Short: 5-15 minutes
# Medium: 15-30 minutes
# Long: > 30 minutes
clean_df['duration_category'] = pd.cut(
    clean_df['trip_duration_minutes'],
    bins=[-np.inf, 5, 15, 30, np.inf],
    labels=['Very Short', 'Short', 'Medium', 'Long']
)

print(clean_df[['trip_duration_minutes',
                'duration_category']].head())
```

```python
# Feature 9 - Route

clean_df['route'] = (
    clean_df['PULocationID'].astype(str)
    + ' → ' +
    clean_df['DOLocationID'].astype(str)
)

print(clean_df[['PULocationID',
                'DOLocationID',
                'route']].head())
```

### Feature Engineering Summary

Created useful features for **trip duration, pickup time, weekdays/weekends, average speed, distance, duration, and routes** to support further analysis.

#Outlier Analysis Using IQR

####Trip Distance IQR

```python
print("IQR Outlier Analysis - Trip Distance")

distance = clean_df['trip_distance'].dropna()

Q1 = np.percentile(distance, 25)
Q3 = np.percentile(distance, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers_distance = clean_df[(clean_df['trip_distance'] < lower_bound) |(clean_df['trip_distance'] > upper_bound)]

print("\nNumber of potential outliers:", len(outliers_distance))
print('-'*100)
print("\nHighest Trip Distance Records")

print(outliers_distance[['trip_distance', 'trip_duration_minutes', 'PULocationID', 'DOLocationID']].sort_values('trip_distance', ascending=False).head(10))
print('-'*100)
print("Conclusion:")
print("Some extreme trip distance values appear to be data errors.")
print("They are paired with very short trip durations, so they require further investigation.")
```

####Trip Duration IQR

```python
print("IQR Outlier Analysis - Trip Duration")

duration = clean_df['trip_duration_minutes'].dropna()

Q1 = np.percentile(duration, 25)
Q3 = np.percentile(duration, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers_duration = clean_df[
    (clean_df['trip_duration_minutes'] < lower_bound) |
    (clean_df['trip_duration_minutes'] > upper_bound)
]

print("\nNumber of potential outliers:", len(outliers_duration))
print('-'*100)
print("\nHighest Trip Duration Records")

print(outliers_duration[['trip_duration_minutes', 'trip_distance', 'PULocationID', 'DOLocationID']].sort_values('trip_duration_minutes', ascending=False).head(10))
print('-'*100)
print("Conclusion:")
print("Some trip durations are extremely high compared to their trip distance.")
print("These records appear to be potential data-quality issues and need further investigation.")
```

####Average Speed IQR

```python
print("IQR Outlier Analysis - Average Speed")

speed = clean_df['average_speed_mph'].dropna()

Q1 = np.percentile(speed, 25)
Q3 = np.percentile(speed, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers_speed = clean_df[
    (clean_df['average_speed_mph'] < lower_bound) |
    (clean_df['average_speed_mph'] > upper_bound)
]

print("\nNumber of potential outliers:", len(outliers_speed))
print('-'*100)
print("Highest Average Speed Records")

print(outliers_speed[['average_speed_mph', 'trip_distance', 'trip_duration_minutes','PULocationID', 'DOLocationID']].sort_values('average_speed_mph', ascending=False).head(10))

print('-'*100)
print("Conclusion:")
print("Some average speed values are extremely high and unrealistic.")
print("These records are likely linked to incorrect distance or duration values and need further investigation.")
```

###IQR Outlier Analysis Conclusion:
----
- The analysis identified extreme values in trip distance, trip duration and average speed.
- Several records appear to be data-quality issues, especially those with unrealistic distance, duration or speed values.
- These records should be investigated rather than automatically deleted.
